```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
det = np.linalg.det(A)
print("Determinant:", det)
```

```python
eigvals, eigvecs = np.linalg.eig(A)
print("Eigenvalues:", eigvals)
print("Eigenvectors:\n", eigvecs)
```

```python
v = np.array([3, 4])
norm = np.linalg.norm(v)
print("Vector Norm:", norm)
```

```python
import numpy as np

# Original feature: age in years
age = np.array([20, 30, 40])
# Feature engineering: age squared
age_squared = np.power(age, 2)
print("Age Squared:", age_squared)
```
